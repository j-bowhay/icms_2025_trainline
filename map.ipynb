```python
import folium

# Create a base map centered on the UK
m = folium.Map(location=[54.5, -3], zoom_start=6)

# Define cities and their coordinates
cities = {
    "London": [51.5074, -0.1278],
    "Manchester": [53.4808, -2.2426],
    "Edinburgh": [55.9533, -3.1883],
    "Bristol": [51.4545, -2.5879],
}

# Add markers
for city, coord in cities.items():
    folium.Marker(location=coord, popup=city).add_to(m)

# Draw edges between chosen pairs
edges = [("London", "Manchester"), ("Manchester", "Edinburgh")]

for city1, city2 in edges:
    folium.PolyLine(locations=[cities[city1], cities[city2]], color='blue').add_to(m)

# Save or display the map
m.save("uk_city_edges.html")
```
